###  Working with Pandas Series in Python

In this example, we will create a **Pandas Series** from a Python list.  
A `Series` is a one-dimensional labeled array capable of holding any data type (integers, floats, strings, etc.).


```python
# Import the required library
import pandas as pd

# Initialize a Python list with mixed data types
data = [1, 2.3, 'a', 4, 5]

# Create a Pandas Series from the list
series_from_list = pd.Series(data)

# Display the Series
print(series_from_list)
```

###  Creating a DataFrame in Pandas

A **DataFrame** is a two-dimensional, size-mutable, and heterogeneous data structure in Pandas.  
It is similar to a table in relational databases or an Excel spreadsheet, where data is aligned in rows and columns.

In this example, we will create a DataFrame using a **Python dictionary**.


```python
# Import the required library
import pandas as pd

# Define data using a Python dictionary
# Each key becomes a column name, and the corresponding list represents the column values
data = {
    'Name': ['praveen', 'Gaurav', 'priyanjilli', 'krishna'],
    'Age': [23, 25, 22, 24],
    'Score': [85, 90, 78, 92]
}
df = pd.DataFrame(data)
print(df)
```

###  Alignment in Pandas Series

When performing arithmetic operations between two Pandas Series,  
Pandas **automatically aligns the data based on the index labels** rather than the position of the elements.

If an index label is missing in one Series, the result will contain **NaN (Not a Number)** for that index.


```python
import pandas as pd

# Create the first Series with custom index labels
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])

# Create the second Series with overlapping and non-overlapping index labels
s2 = pd.Series([4, 5, 6], index=["b", "c", "d"])

# Perform element-wise multiplication
# Pandas aligns values based on matching index labels
result = s1 * s2

print(result)
```

```python
#alignment 
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([4, 5, 6], index=["b", "c", "d"])
print(s1 * s2)
```

###  Arithmetic Operations on Pandas Series

When two **Pandas Series** have the same default integer index,  
arithmetic operations are performed **element-wise by position**.

```python
import pandas as pd

# Create the first Series with default integer index (0, 1, 2)
series_a = pd.Series([1, 2, 3])

# Create the second Series with the same default index
series_b = pd.Series([4, 5, 6])

# Perform element-wise addition of the two Series
sum_series = series_a + series_b

# Display the result
print(sum_series)
```

###  Creating a MultiIndex Series in Pandas

A **MultiIndex Series** allows us to have multiple levels of indexing (hierarchical indexing).  
This is useful when working with data that has more than one categorical key,  
such as grouping by *Category* and *Subcategory*.


```python
# Import the Pandas library
import pandas as pd

# Define two arrays to represent the hierarchical index
arrays = [
    ['A', 'A', 'B', 'B'],             # First level index
    ['Math', 'Science', 'Math', 'Science']  # Second level index
]

# Create a MultiIndex from the arrays with names for each level
index = pd.MultiIndex.from_arrays(arrays, names=('Alphabet', 'Subject'))

# Create a Series with MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

print(multi_s)
```

###  Creating a MultiIndex Series from Tuples

Another way to create a **MultiIndex Series** in Pandas is by using a list of **tuples**,  
where each tuple represents a combination of index values.  

This approach is often more compact and intuitive compared to defining separate arrays.


```python
# Import the Pandas library
import pandas as pd  

# Define a list of tuples where each tuple represents (Alphabet, Subject)
tuples = [
    ('A', 'Math'),
    ('A', 'Science'),
    ('B', 'Math'),
    ('B', 'Science')
]

# Create a MultiIndex from the list of tuples and assign level names
index = pd.MultiIndex.from_tuples(tuples, names=('Alphabet', 'Subject'))

# Create a Series with MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

print(multi_s)
```

###  Creating a MultiIndex Series using `from_product`

Pandas provides `pd.MultiIndex.from_product()` to create a **cartesian product** of iterables.  
This is useful when you want **all possible combinations** of multiple index levels.  

In this example, we create a MultiIndex Series of student scores  
with two levels: **Alphabet** and **Subject**.


```python
# Import the Pandas library
import pandas as pd

# Create a MultiIndex using cartesian product of two lists
index = pd.MultiIndex.from_product(
    [['A', 'B'], ['Math', 'Science']],   # All possible combinations
    names=('Alphabet', 'Subject')        # Assign level names
)

# Create a Series with MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

# Display the MultiIndex Series
print(multi_s)
```

###  Creating a MultiIndex Series from a DataFrame

Pandas allows you to create a **MultiIndex** from an existing DataFrame using `from_frame()`.  
This is useful when you already have structured tabular data and want to use its columns as hierarchical index levels.

In this example, we create a MultiIndex Series of student scores based on two columns: **Alphabet** and **Subject**.


```python
import pandas as pd

# Create a DataFrame with columns representing hierarchical index levels
df = pd.DataFrame({
    'Alphabet': ['A', 'A', 'B', 'B'],
    'Subject': ['Math', 'Science', 'Math', 'Science']
})

# Create a MultiIndex from the DataFrame
index = pd.MultiIndex.from_frame(df, names=('Alphabet', 'Subject'))

# Create a Series with MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

print(multi_s)
```

###  MultiIndex in Pandas

This notebook demonstrates **advanced usage of MultiIndex** in Pandas, including:

1. Creating MultiIndex Series in multiple ways  
2. Accessing, slicing, and indexing  
3. Swapping and reordering levels  
4. Automatic MultiIndex construction  
5. Data alignment and reindexing  
6. Using `xs()` for cross-section  
7. Sorting and removing unused levels


#### 1 Creating a MultiIndex Series in Different Ways

We can create MultiIndex Series from:

- Arrays  
- Tuples  
- Cartesian product of iterables (`from_product`)


```python
import pandas as pd
import numpy as np

# ----- From arrays -----
arrays = [["A", "A", "B", "B"], ["Math", "Science", "Math", "Science"]]
index = pd.MultiIndex.from_arrays(arrays, names=("Alphabet", "Subject"))
multi_s = pd.Series([90, 85, 88, 92], index=index)
print("MultiIndex Series from arrays:\n", multi_s, "\n")

# ----- From tuples -----
tuples = [("A", "Math"), ("A", "Science"), ("B", "Math"), ("B", "Science")]
index2 = pd.MultiIndex.from_tuples(tuples, names=("Alphabet", "Subject"))
multi_s2 = pd.Series([70, 75, 80, 82], index=index2)
print("MultiIndex Series from tuples:\n", multi_s2, "\n")

# ----- From product (Cartesian product) -----
iterables = [["A", "B"], ["Math", "Science"]]
index3 = pd.MultiIndex.from_product(iterables, names=("Alphabet", "Subject"))
multi_s3 = pd.Series(np.random.randint(60, 100, size=4), index=index3)
print("MultiIndex Series from product:\n", multi_s3, "\n")
```

#### 2 Accessing and Indexing

We can access data using `.loc` and provide:

- A single index level  
- Tuple for multiple levels


```python
# Access all subjects for 'A'
print("All subjects for 'A':\n", multi_s.loc["A"], "\n")

# Access specific element
print("Specific element (B, Science):\n", multi_s.loc[("B", "Science")], "\n")
```

#### 3 Slicing in MultiIndex

- Slice by index levels  
- Partial slices across levels

```python
# Slice from 'A' to 'B'
print("Slice from A to B:\n", multi_s.loc["A":"B"], "\n")

# Partial slice for all Math subjects
print("Partial slice for all Math:\n", multi_s.loc[:, "Math"], "\n")
```

#### 4 Swapping and Reordering Levels

```python
# Swap levels
print("Swapped levels:\n", multi_s.swaplevel(), "\n")

# Reorder levels
print("Reordered levels:\n", multi_s3.reorder_levels(["Subject", "Alphabet"]), "\n")
```

#### 5 Passing List of Arrays Directly to Series / DataFrame

```python
# MultiIndex Series automatically
multi_s_auto = pd.Series(np.random.randn(4),
                         index=pd.MultiIndex.from_arrays([["A", "A", "B", "B"], ["X", "Y", "X", "Y"]]))
print("MultiIndex Series constructed automatically:\n", multi_s_auto, "\n")

# MultiIndex DataFrame automatically
df_auto = pd.DataFrame(np.random.randn(4, 2),
                       index=pd.MultiIndex.from_arrays([["Group1", "Group1", "Group2", "Group2"], ["One", "Two", "One", "Two"]]),
                       columns=["Score1", "Score2"])
print("DataFrame with MultiIndex automatically:\n", df_auto, "\n")
```

#### 6 Data Alignment and Reindexing

```python
# Original DataFrame with MultiIndex
df = pd.DataFrame({"Math": [85, 90, 95, 80], "Science": [82, 88, 92, 84]},
                  index=pd.MultiIndex.from_arrays([["A", "A", "B", "B"], ["one", "two", "one", "two"]]))
print("Original DataFrame:\n", df, "\n")

# Group by first level and compute mean
mean_by_group = df.groupby(level=0).mean()
print("Mean by group:\n", mean_by_group, "\n")

# Reindexing to align with original index
aligned = mean_by_group.reindex(df.index, level=0)
print("Reindexed to align with original index:\n", aligned, "\n")
```

#### 7 Using `xs()` for Cross-Section

```python
# Cross-section for level 'two'
print("Cross-section for level 'two':\n", df.xs("two", level=1), "\n")
```

#### 8 Sorting and Removing Unused Levels

```python
# Create a DataFrame with MultiIndex columns
arrays = [["Group1", "Group1", "Group2", "Group2"], ["One", "Two", "One", "Two"]]
columns = pd.MultiIndex.from_arrays(arrays, names=("Group", "Subgroup"))

df_multi = pd.DataFrame(np.random.randn(4, 4), columns=columns)
print("Original MultiIndex DataFrame:\n", df_multi, "\n")

# Drop a column to create unused levels
df_sub = df_multi.drop(columns=[("Group2", "Two")])
print("After dropping a column:\n", df_sub, "\n")

# Remove unused levels
df_cleaned = df_sub.copy()
df_cleaned.columns = df_cleaned.columns.remove_unused_levels()
print("After removing unused levels:\n", df_cleaned.columns.levels, "\n")
```
