### Part 1: Working with CSV in Python


In this section, we will learn everything about handling **CSV (Comma-Separated Values)** files using Python.  
We will cover:
- What CSV files are
- Reading CSV files
- Writing CSV files
- Appending to CSV files
- Using `csv` module
- Using `pandas` for advanced operations

CSV is one of the most widely used file formats for **data exchange**.


```python
# Importing necessary libraries
import csv   # Built-in Python module for handling CSV files
import pandas as pd  # Popular library for advanced CSV handling
```

### Creating a Sample CSV File
Before learning to read and write, let's create a sample CSV file for practice.


```python
# Creating a sample CSV file
sample_data = [
    ["ID", "Name", "Age", "Department"],
    [1, "praveen", 24, "HR"],
    [2, "gaurav", 27, "Finance"],
    [3, "priyanjali", 22, "IT"],
    [4, "krishna", 29, "Marketing"]
]

# Writing data into a CSV file
with open("employees.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(sample_data)

print("Sample CSV file 'employees.csv' created successfully.")
```

### Reading CSV using `csv.reader`
The `csv.reader` object allows us to read rows line by line from a CSV file.

```python
# Reading CSV file using csv.reader
with open("employees.csv", mode="r") as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        print(row)
```

### Reading CSV using `csv.DictReader`
- Reads the CSV as a list of dictionaries.  
- Each row is represented as a dictionary where **keys = column headers**.


```python
# Reading CSV using DictReader
with open("employees.csv", mode="r") as file:
    dict_reader = csv.DictReader(file)
    for row in dict_reader:
        print(row)   # Each row is a dictionary
```

### Writing CSV using `csv.writer`
We can write data to a new CSV file using `writerow()` or `writerows()`.


```python
# Writing data into a new CSV file
new_data = [
    ["ID", "Name", "Score"],
    [1, "praveen", 90],
    [2, "gaurav", 85],
    [3, "priyanjali", 88]
]

with open("scores.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(new_data)

print("New CSV file 'scores.csv' created successfully.")
```

### Appending Data to CSV
Use mode `"a"` (append) to add new rows to an existing CSV file.

```python
# Appending data to existing CSV
with open("scores.csv", mode="a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow([4, "krishna", 92])

print("Data appended successfully to 'scores.csv'.")
```

### Reading CSV using Pandas
Pandas provides a powerful function `pd.read_csv()` to load CSV files into DataFrames.


```python
# Reading CSV file using pandas
df = pd.read_csv("employees.csv")
print(df)

# Display only column names
print("\nColumns in CSV:", df.columns.tolist())
```

### Writing CSV using Pandas
We can export DataFrames into CSV using `to_csv()`.


```python
# Writing DataFrame to CSV
df.to_csv("employees_copy.csv", index=False)
print("DataFrame exported as 'employees_copy.csv'")
```

### Selecting Specific Columns and Saving
We can filter columns and save them into a new CSV file.

```python
# Selecting only "Name" and "Department" columns
df_filtered = df[["Name", "Department"]]
df_filtered.to_csv("employee_departments.csv", index=False)

print("Filtered CSV 'employee_departments.csv' created successfully.")
```

### Advanced CSV Handling with Pandas
Some useful options in `read_csv()`:
- `sep` → Change delimiter (e.g., `sep=";"`)
- `header` → Specify header row
- `nrows` → Read only first n rows
- `usecols` → Load specific columns
- `skiprows` → Skip certain rows


### Example: Using `sep`
By default, CSV uses **comma (`,`)** as a delimiter.  
But sometimes files use `;`, `|`, `\t` (tab), etc.  
We can specify the delimiter with `sep`.


```python
# Creating a CSV file with semicolon as delimiter
data = "ID;Name;Age\n1;Alice;24\n2;Bob;27\n3;Charlie;22"
with open("semicolon_employees.csv", "w") as f:
    f.write(data)

# Reading CSV with custom delimiter
df_sep = pd.read_csv("semicolon_employees.csv", sep=";")
print(df_sep)
```

### Example: Using `header`
- The `header` parameter specifies which row should be used as column names.
- `header=0` → First row is header (default).
- `header=None` → No header; assigns default column names as `0,1,2,...`.
- `header=1` → Second row is used as header.


```python
# Creating a CSV with an extra row before actual headers
data = "This is employee data\nID,Name,Age\n1,Alice,24\n2,Bob,27"
with open("header_example.csv", "w") as f:
    f.write(data)

# Case 1: Default (header=0 → first row is header)
# Pandas incorrectly treats "This is employee data" as header
df_header_default = pd.read_csv("header_example.csv")
print("Default header:\n", df_header_default)

# Case 2: No header (header=None) → skip the first line since it has fewer columns
# This assigns default column names (0,1,2,...)
df_no_header = pd.read_csv("header_example.csv", header=None, skiprows=1)
print("\nNo header:\n", df_no_header)

# Case 3: Use second row as header (skip the first line and then set header=0)
# Correct way → "ID,Name,Age" becomes header row
df_second_header = pd.read_csv("header_example.csv", header=0, skiprows=1)
print("\nSecond row as header:\n", df_second_header)
```

### Example: Using `nrows`
- The `nrows` parameter limits how many rows of data are read.
- Useful for previewing large CSV files.


```python
# Reading only the first 2 rows
df_nrows = pd.read_csv("employees.csv", nrows=2)
print(df_nrows)
```

### Example: Using `usecols`
- The `usecols` parameter loads **only specific columns**.
- Saves memory and speeds up processing.

```python
# Reading only specific columns
df_usecols = pd.read_csv("employees.csv", usecols=["Name", "Department"])
print(df_usecols)
```

### Example: Using `skiprows`
- The `skiprows` parameter skips specified rows.
- Can pass:
  - An integer → number of rows to skip
  - A list of row indices → skip specific rows


```python
# Case 1: Skip first row
df_skip_1 = pd.read_csv("employees.csv", skiprows=1)
print("Skipping first row:\n", df_skip_1)

# Case 2: Skip multiple specific rows
df_skip_list = pd.read_csv("employees.csv", skiprows=[1, 3])
print("\nSkipping rows 1 and 3:\n", df_skip_list)
```

### Part 2: Working with JSON in Python



In this section, we will learn about **JSON (JavaScript Object Notation)**.  

- JSON is a lightweight data-interchange format.  
- It is easy for humans to read and write.  
- It is easy for machines to parse and generate.  
- Commonly used in APIs, web applications, and data storage.

We will cover:
- Creating JSON data
- Reading JSON (from string and file)
- Writing JSON to file
- Working with nested JSON
- Using Python `json` module
- Using Pandas with JSON


```python
import json   # Built-in module for JSON handling
import pandas as pd  # For advanced JSON operations
```

### Creating JSON Data
JSON data in Python is represented as a dictionary or list.


```python
# Creating a Python dictionary (JSON-like structure)
employee = {
    "id": 1,
    "name": "praveen",
    "age": 24,
    "department": "HR",
    "skills": ["Data analyst", "Software developer"]
}

print(employee)
```

### Converting Python Object to JSON String
Use `json.dumps()` to convert Python object into a JSON-formatted string.


```python
# Convert dictionary to JSON string
employee_json = json.dumps(employee, indent=4)
print(employee_json)
```

### Writing JSON Data to a File
Use `json.dump()` to write JSON directly to a file.


```python
# Writing JSON to file
with open("employee.json", "w") as file:
    json.dump(employee, file, indent=4)

print("JSON data written to 'employee.json'")
```

### Reading JSON String
Use `json.loads()` to convert JSON string back into a Python object.


```python
# JSON string
json_string = '{"id": 2, "name": "gaurav", "age": 27, "department": "Finance"}'

# Convert JSON string to Python dictionary
employee_obj = json.loads(json_string)
print(employee_obj)
```

### Reading JSON Data from File
Use `json.load()` to read JSON from a file.


```python
# Reading JSON file
with open("employee.json", "r") as file:
    employee_data = json.load(file)

print(employee_data)
```

### Working with Nested JSON
JSON can be nested (contain objects and arrays inside).


```python
# Nested JSON
company = {
    "company": "TechCorp",
    "employees": [
        {"id": 1, "name": "praveen", "role": "HR"},
        {"id": 2, "name": "gaurav", "role": "Finance"},
        {"id": 3, "name": "priyanjali", "role": "IT"}
    ]
}

# Accessing nested data
print("Company Name:", company["company"])
print("First Employee:", company["employees"][0]["name"])
```

```python
# Save nested JSON into a file
with open("company.json", "w") as file:
    json.dump(company, file, indent=4)

print("Nested JSON written to 'company.json'")
```

### Reading JSON with Pandas
Pandas can directly read JSON into a DataFrame using `pd.read_json()`.


```python
# Read JSON file into DataFrame
df_json = pd.read_json("company.json")
print(df_json)

# Normalize nested JSON into a flat table
df_normalized = pd.json_normalize(company["employees"])
print("\nNormalized Employee Data:\n", df_normalized)
```

### Part 3: Working with Excel in Python


In this section, we will learn how to handle **Excel files (.xlsx)** in Python.  

We will cover:
- Reading Excel files
- Writing Excel files
- Appending data
- Reading specific sheets
- Reading multiple sheets
- Writing multiple sheets
- Using `openpyxl` for advanced operations


```python
import pandas as pd   # Best library for Excel handling
from openpyxl import Workbook, load_workbook  # For advanced Excel operations
```

```python
# Creating a sample DataFrame
data = {
    "ID": [1, 2, 3, 4],
    "Name": ["praveen", "Gaurav", "priyanjali", "krishna"],
    "Age": [24, 27, 22, 29],
    "Department": ["HR", "Finance", "IT", "Marketing"]
}
df = pd.DataFrame(data)

# Saving DataFrame to Excel
df.to_excel("employees.xlsx", index=False, sheet_name="EmployeeData")

print("Sample Excel file 'employees.xlsx' created successfully.")
```

### Reading Excel File
Use `pd.read_excel()` to read Excel files.


```python
# Reading Excel file
df_excel = pd.read_excel("employees.xlsx")
print(df_excel)
```

### Reading Specific Sheet
If the Excel file has multiple sheets, specify sheet name using `sheet_name`.


```python
# Reading specific sheet
df_sheet = pd.read_excel("employees.xlsx", sheet_name="EmployeeData")
print(df_sheet)
```

### Writing Multiple Sheets
We can write DataFrames into multiple sheets using `ExcelWriter`.


```python
# Creating multiple DataFrames
df_scores = pd.DataFrame({
    "ID": [1, 2, 3],
    "Score": [90, 85, 88]
})

# Writing multiple sheets
with pd.ExcelWriter("company_data.xlsx") as writer:
    df.to_excel(writer, sheet_name="Employees", index=False)
    df_scores.to_excel(writer, sheet_name="Scores", index=False)

print("Data written to multiple sheets in 'company_data.xlsx'")
```

### Reading Multiple Sheets
We can read all sheets at once into a dictionary of DataFrames.


```python
# Reading all sheets
all_sheets = pd.read_excel("company_data.xlsx", sheet_name=None)
for sheet, data in all_sheets.items():
    print(f"Sheet: {sheet}\n", data, "\n")
```

### Appending Data to Excel
Pandas `to_excel()` overwrites files by default.  
To append, we use **openpyxl**.


```python
# Load existing workbook
wb = load_workbook("employees.xlsx")
ws = wb.active

# Append new row
ws.append([5, "kumar", 26, "Sales"])
wb.save("employees.xlsx")

print("New row appended successfully to 'employees.xlsx'")
```

### Advanced Options with Pandas
- `usecols` → Read specific columns
- `skiprows` → Skip unwanted rows
- `nrows` → Read limited rows


```python
# Reading only specific columns
df_usecols = pd.read_excel("employees.xlsx", usecols=["Name", "Department"])
print("Specific columns:\n", df_usecols)

# Reading limited rows
df_nrows = pd.read_excel("employees.xlsx", nrows=2)
print("\nOnly first 2 rows:\n", df_nrows)
```

### Styling Excel Sheets (with openpyxl)
We can apply styling (fonts, colors, bold, etc.) to Excel cells.


```python
from openpyxl.styles import Font

# Load workbook
wb = load_workbook("employees.xlsx")
ws = wb.active

# Make header row bold
for cell in ws[1]:
    cell.font = Font(bold=True, color="0000FF")

wb.save("employees_styled.xlsx")
print("Styled Excel file 'employees_styled.xlsx' created successfully.")
```
