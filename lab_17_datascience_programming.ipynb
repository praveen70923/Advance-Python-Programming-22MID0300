###  Creating a Pandas Series from a List

A **Pandas Series** is a one-dimensional labeled array capable of holding data of any type — integers, floats, or strings.  
It is similar to a single column in an Excel sheet or a database table.

In this example, we will create a **Series** using a Python list.


```python
import pandas as pd

# Define a list containing mixed data types (integers, float, string)
data = [1, 2.3, 'a', 4, 5]

# Create a Pandas Series from the list
series_from_list = pd.Series(data)

# Display the created Series
print(series_from_list)
```

### Creating a Pandas Series from a List
This code demonstrates how to create a Pandas Series using a Python list that contains mixed data types such as integers, floats, and strings.


```python
import pandas as pd

# Define a list containing mixed data types (integers, float, string)
data = [1, 2.3, 'a', 4, 5]

# Create a Pandas Series from the list
series_from_list = pd.Series(data)

# Display the created Series
print(series_from_list)
```

### Series Alignment During Arithmetic Operations

When performing arithmetic operations between two Pandas Series, alignment is done based on the index labels.  
If an index is missing in either Series, the result will contain `NaN` for that index.


```python
# Create two Series with partially overlapping indexes
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([4, 5, 6], index=["b", "c", "d"])

# Multiply the Series (alignment based on index)
print(s1 * s2)
```

### Element-wise Operations Between Series

When two Pandas Series share the same index, arithmetic operations like addition, subtraction, etc.,  
are performed **element-wise** on corresponding elements.


```python
series_a = pd.Series([1, 2, 3])
series_b = pd.Series([4, 5, 6])

# Perform element-wise addition
sum_series = series_a + series_b

print(sum_series)
```

### Creating a MultiIndex Series

A **MultiIndex Series** allows hierarchical indexing, enabling data to be organized in multiple dimensions.  
This example demonstrates creating a MultiIndex Series using `from_arrays()` — where two lists define two index levels.


```python
# Creating a MultiIndex Series
arrays = [
    ['A', 'A', 'B', 'B'],                # First level index
    ['Math', 'Science', 'Math', 'Science']  # Second level index
]
index = pd.MultiIndex.from_arrays(arrays, names=('Alphabet', 'Subject'))

multi_s = pd.Series([90, 85, 88, 92], index=index)
print(multi_s)
```

### Creating a MultiIndex Series using `from_arrays()`

A **MultiIndex Series** in Pandas allows representation of data with more than one index level.  
Here, we define two arrays — one for alphabets and another for subjects — and use them to create a hierarchical index.


```python
# Creating a MultiIndex Series
import pandas as pd  

arrays = [
    ['A', 'A', 'B', 'B'],                 # First-level index (Alphabet)
    ['Math', 'Science', 'Math', 'Science']  # Second-level index (Subject)
]

# Create MultiIndex from the arrays
index = pd.MultiIndex.from_arrays(arrays, names=('Alphabet', 'Subject'))

# Create Series with the defined MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

# Display the Series
print(multi_s)
```

### Creating a MultiIndex Series using `from_arrays()`

A **MultiIndex Series** in Pandas allows representation of data with more than one index level.  
Here, we define two arrays — one for alphabets and another for subjects — and use them to create a hierarchical index.


```python
# Creating a MultiIndex Series using Cartesian product
import pandas as pd

# Create MultiIndex from product of two lists
index = pd.MultiIndex.from_product(
    [['A', 'B'], ['Math', 'Science']],  # All possible combinations
    names=('Alphabet', 'Subject')       # Assign level names
)

# Create Series with MultiIndex
multi_s = pd.Series([90, 85, 88, 92], index=index)

# Display the Series
print(multi_s)
```

### Exploring MultiIndex Series and DataFrames in Pandas

This section demonstrates the creation, manipulation, and indexing of **MultiIndex Series** and **DataFrames**.  
A **MultiIndex** (hierarchical index) allows you to represent higher-dimensional data in a 1D or 2D structure, enabling powerful data organization and analysis.

---

### 1. Creating a MultiIndex Series in Different Ways
We can create a MultiIndex Series using:
- **from_arrays()** – combining multiple index arrays  
- **from_tuples()** – list of tuples as index pairs  
- **from_product()** – Cartesian product of lists


```python
import pandas as pd
import numpy as np

# From arrays
arrays = [["A", "A", "B", "B"], ["Math", "Science", "Math", "Science"]]
index = pd.MultiIndex.from_arrays(arrays, names=("Alphabet", "Subject"))
multi_s = pd.Series([90, 85, 88, 92], index=index)
print("MultiIndex Series from arrays:\n", multi_s, "\n")

# From tuples
tuples = [("A", "Math"), ("A", "Science"), ("B", "Math"), ("B", "Science")]
index2 = pd.MultiIndex.from_tuples(tuples, names=("Alphabet", "Subject"))
multi_s2 = pd.Series([70, 75, 80, 82], index=index2)
print("MultiIndex Series from tuples:\n", multi_s2, "\n")

# From product
iterables = [["A", "B"], ["Math", "Science"]]
index3 = pd.MultiIndex.from_product(iterables, names=("Alphabet", "Subject"))
multi_s3 = pd.Series(np.random.randint(60, 100, size=4), index=index3)
print("MultiIndex Series from product:\n", multi_s3, "\n")
```

### 2. Accessing and Indexing Elements
We can access subsets or individual elements using `.loc[]`.


```python
print("Access all subjects for 'A':\n", multi_s.loc["A"], "\n")
print("Access specific element (B, Science):\n", multi_s.loc[('B', 'Science')], "\n")
```

### 3. Slicing in MultiIndex
You can perform slicing operations across levels using `.loc[]`.


```python
print("Slicing from A to B:\n", multi_s.loc["A":"B"], "\n")
print("Partial slice for all Math:\n", multi_s.loc[:, "Math"], "\n")
```

### 4. Swapping and Reordering Levels
Levels of a MultiIndex can be **swapped** or **reordered** to change the data view.


```python
print("Swapping levels:\n", multi_s.swaplevel(), "\n")
print("Reordering levels:\n", multi_s3.reorder_levels(["Subject", "Alphabet"]), "\n")
```

### 5. Constructing MultiIndex Automatically
We can directly pass lists of arrays to Series or DataFrame constructors.


```python
multi_s_auto = pd.Series(
    np.random.randn(4),
    index=pd.MultiIndex.from_arrays([["A", "A", "B", "B"], ["X", "Y", "X", "Y"]])
)
print("MultiIndex Series constructed automatically:\n", multi_s_auto, "\n")

df_auto = pd.DataFrame(
    np.random.randn(4, 2),
    index=pd.MultiIndex.from_arrays([["Group1", "Group1", "Group2", "Group2"], ["One", "Two", "One", "Two"]]),
    columns=["Score1", "Score2"]
)
print("DataFrame with MultiIndex automatically:\n", df_auto, "\n")
```

### 6. Data Alignment and Reindexing
We can align and reindex data efficiently using MultiIndex structures.


```python
df = pd.DataFrame({
    "Math": [85, 90, 95, 80],
    "Science": [82, 88, 92, 84]
}, index=pd.MultiIndex.from_arrays([["A", "A", "B", "B"], ["one", "two", "one", "two"]]))
print("Original DataFrame:\n", df, "\n")

mean_by_group = df.groupby(level=0).mean()
print("Mean by group:\n", mean_by_group, "\n")

aligned = mean_by_group.reindex(df.index, level=0)
print("Reindexed to align with original index:\n", aligned, "\n")
```

### 7. Using `xs()` for Cross-Section
The `xs()` method allows quick selection of data across levels.


```python
print("Cross-section for level 'two':\n", df.xs("two", level=1), "\n")
```

### 8. Sorting and Removing Unused Levels
Sorting organizes the data by index, while `remove_unused_levels()` cleans up unused MultiIndex levels.


```python
# -----------------------------------------------------------
# 8. Sorting and Removing Unused Levels (Fixed)
# -----------------------------------------------------------

# Shuffle and sort MultiIndex Series
unsorted = multi_s_auto.sample(frac=1)
print("Unsorted MultiIndex Series:\n", unsorted, "\n")
print("Sorted by index:\n", unsorted.sort_index(), "\n")

# Create a DataFrame with MultiIndex columns
arrays = [["Score", "Score"], ["Math", "Science"]]
columns = pd.MultiIndex.from_arrays(arrays, names=["Category", "Subject"])

df_multi_col = pd.DataFrame(
    np.random.randint(70, 100, size=(4, 2)),
    index=pd.Index(["A", "B", "C", "D"], name="Alphabet"),
    columns=columns
)

print("DataFrame with MultiIndex columns:\n", df_multi_col, "\n")

# Drop one column and clean unused levels
sub_df = df_multi_col[["Score", "Math"]]
print("Before removing unused levels:\n", sub_df.columns.levels, "\n")
sub_df.columns = sub_df.columns.remove_unused_levels()
print("After removing unused levels:\n", sub_df.columns.levels, "\n")
```
