#### 1. Introduction to Matplotlib
Matplotlib is the most widely used Python library for creating 2D plots and visualizations.  
It provides a MATLAB-like interface (`pyplot`) for quick plotting and also supports object-oriented plotting for more control.  

Common uses:
- Visualizing data distributions
- Comparing trends
- Building charts for data analysis


```python
# Importing Matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Simple data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)  # Line plot
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4])
plt.ylabel('some numbers')
plt.show()
```

#### 2. Line Plot (plt.plot)
A line plot shows data points connected by a straight line.  
Useful for showing trends over time or continuous variables.

```python
x = np.linspace(0, 10, 100)   # 100 points between 0 and 10
y = np.sin(x)                 # Sine function

plt.plot(x, y, color='blue', linestyle='--', marker='o', label='sin(x)')
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

#### 3. Scatter Plot (plt.scatter)
A scatter plot shows individual data points without connecting lines.  
Useful for identifying relationships between two variables.

```python
np.random.seed(42)
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='red', marker='x')
plt.title("Scatter Plot Example")
plt.xlabel("Random X")
plt.ylabel("Random Y")
plt.show()
```

```python
import numpy as np

# evenly sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```

This example demonstrates plotting multiple datasets with **different line styles and markers** in a single plot.  

- `t = np.arange(0., 5., 0.2)` → Creates an array of values from 0 to 5 with step 0.2.  
- `'r--'` → Red dashed line.  
- `'bs'` → Blue squares as markers.  
- `'g^'` → Green triangles as markers.  
- `plt.plot()` can take multiple `(x, y, format)` sets to plot multiple lines at once.  
- `plt.show()` → Displays the plot.  


```python
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

plt.scatter('a', 'b', c='c', s='d', data=data)
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

This example demonstrates a **scatter plot** in Matplotlib with variable **point colors** and **sizes**.  

- `data['a']` → X-axis values.  
- `data['b']` → Y-axis values.  
- `c='c'` → Colors of points mapped from column `'c'`.  
- `s='d'` → Sizes of points mapped from column `'d'`.  
- `plt.scatter(x, y, c=..., s=..., data=...)` → Creates the scatter plot using data from a dictionary or DataFrame.  
- `plt.xlabel()` and `plt.ylabel()` → Label the axes.  
- `plt.show()` → Displays the plot.  


```python
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100]

plt.figure(figsize=(9, 3))

plt.subplot(131)
plt.bar(names, values)
plt.subplot(132)
plt.scatter(names, values)
plt.subplot(133)
plt.plot(names, values)
plt.suptitle('Categorical Plotting')
plt.show()
```

This example demonstrates **categorical plotting** in Matplotlib using the same dataset for three types of plots.  

- `names` → Categorical labels (X-axis).  
- `values` → Numeric values (Y-axis).  
- `plt.figure(figsize=(9, 3))` → Sets the figure size.  
- `plt.subplot(1, 3, i)` → Creates a 1x3 grid of subplots and selects the i-th subplot.  
- `plt.bar(names, values)` → Bar chart of categorical data.  
- `plt.scatter(names, values)` → Scatter plot of categorical data.  
- `plt.plot(names, values)` → Line plot of categorical data.  
- `plt.suptitle()` → Adds a main title for the figure.  
- `plt.show()` → Displays all subplots.

```python
# Sample data
x1 = np.linspace(0, 5, 100)
y1 = np.sin(x1)
x2 = np.linspace(0, 5, 100)
y2 = np.cos(x2)

# Plot multiple lines
lines = plt.plot(x1, y1, x2, y2)

# Customize lines using keyword arguments
plt.setp(lines, color='r', linewidth=2.0)

# Or using MATLAB-style string-value pairs
plt.setp(lines, 'color', 'r', 'linewidth', 2.0)

plt.title("Multiple Lines with Custom Styling")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

`plt.setp()` allows setting **properties for one or more Matplotlib objects** at once, similar to MATLAB style syntax.  

- `lines = plt.plot(x1, y1, x2, y2)` → Plots multiple lines and returns a list of `Line2D` objects.  
- `plt.setp(lines, color='r', linewidth=2.0)` → Sets the color and line width for all lines in the list.  
- Alternative MATLAB-style syntax: `plt.setp(lines, 'color', 'r', 'linewidth', 2.0)`.  
- Useful for efficiently applying the same styling to multiple lines.


```python
def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure()
plt.subplot(211)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), 'r--')
plt.show()
```

This example demonstrates **creating multiple subplots** in a single figure to compare different datasets.  

- `def f(t)` → Defines a function `f(t) = exp(-t) * cos(2πt)`.  
- `t1` and `t2` → Two time arrays with different sampling intervals.  
- `plt.figure()` → Creates a new figure.  
- `plt.subplot(211)` → Creates the first subplot in a 2x1 grid (2 rows, 1 column).  
- `plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')` → Plots two datasets in the first subplot:  
  - `'bo'` → Blue circles  
  - `'k'` → Black line  
- `plt.subplot(212)` → Creates the second subplot in the 2x1 grid.  
- `'r--'` → Red dashed line.  
- `plt.show()` → Displays the figure with both subplots.  


```python
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# the histogram of the data
n, bins, patches = plt.hist(x, 50, density=True, facecolor='r', alpha=0.75)


plt.xlabel('Smarts')
plt.ylabel('Probability')
plt.title('Histogram of IQ')
plt.text(60, .025, r'$\mu=100,\ \sigma=15$')
plt.axis([40, 160, 0, 0.03])
plt.grid(True)
plt.show()
```

This example demonstrates creating a **histogram** for normally distributed data with custom styling and annotations.  

- `mu, sigma = 100, 15` → Mean and standard deviation for a normal distribution.  
- `x = mu + sigma * np.random.randn(10000)` → Generates 10,000 random samples.  
- `plt.hist(x, 50, density=True, facecolor='g', alpha=0.75)` →  
  - `50` bins  
  - `density=True` → Normalizes the histogram to show probability density  
  - `facecolor='g'` → Green bars  
  - `alpha=0.75` → Transparency  
- `plt.text()` → Adds LaTeX-style annotation for mean and standard deviation.  
- `plt.axis()` → Sets axis limits.  
- `plt.grid(True)` → Shows grid lines.

```python
plt.xscale('log')
```

Matplotlib allows setting **logarithmic scales** on the x-axis to better visualize data spanning several orders of magnitude.  

- `plt.xscale('log')` → Changes the x-axis to a logarithmic scale.  
- Useful for exponential growth data or data covering multiple magnitudes.  
- Can be combined with `plt.yscale()` for log-log plots.  
- Other options: `'linear'`, `'symlog'`, `'logit'`.  


```python
# Fixing random state for reproducibility
np.random.seed(19680801)

# make up some data in the open interval (0, 1)
y = np.random.normal(loc=0.5, scale=0.4, size=1000)
y = y[(y > 0) & (y < 1)]
y.sort()
x = np.arange(len(y))

# plot with various axes scales
plt.figure()

# linear
plt.subplot(221)
plt.plot(x, y)
plt.yscale('linear')
plt.title('linear')
plt.grid(True)

# log
plt.subplot(222)
plt.plot(x, y)
plt.yscale('log')
plt.title('log')
plt.grid(True)

# symmetric log
plt.subplot(223)
plt.plot(x, y - y.mean())
plt.yscale('symlog', linthresh=0.01)
plt.title('symlog')
plt.grid(True)

# logit
plt.subplot(224)
plt.plot(x, y)
plt.yscale('logit')
plt.title('logit')
plt.grid(True)
# Adjust the subplot layout, because the logit one may take more space
# than usual, due to y-tick labels like "1 - 10^{-3}"
plt.subplots_adjust(top=0.92, bottom=0.08, left=0.10, right=0.95, hspace=0.25,
                    wspace=0.35)

plt.show()
```

Matplotlib allows plotting data with **various y-axis scales** to better visualize different data distributions.  

- `np.random.seed()` → Fixes the random state for reproducibility.  
- `y = np.random.normal(loc=0.5, scale=0.4, size=1000)` → Generates normally distributed data.  
- `y = y[(y>0) & (y<1)]` → Keeps values within (0,1) for log and logit scales.  
- `x = np.arange(len(y))` → X-axis values.  
- `plt.yscale()` → Sets the scale of the y-axis. Available scales:  
  - `'linear'` → Default linear scale  
  - `'log'` → Logarithmic scale  
  - `'symlog'` → Symmetric log scale for data around zero  
  - `'logit'` → Logit scale for data in (0,1) interval  
- `plt.subplots_adjust()` → Adjusts spacing between subplots to avoid overlapping labels.
