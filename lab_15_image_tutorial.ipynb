## PART-A Introduction 

#### Plotting Images with Matplotlib
#### 1. Import Required Libraries

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

To start using Matplotlib, we need to import its modules.  

- `matplotlib.pyplot` → Provides plotting functions similar to MATLAB (commonly imported as `plt`).  
- `matplotlib.image` → Provides functions to read and display images (commonly imported as `mpimg`).  


#### 2. Load an Image

```python
# Load an image from a file
img = mpimg.imread("the-deer-bug-the-stag-beetle-insects-oak.jpg")
```

#### 3. Display the Image

```python
plt.imshow(img)       # Show the image
plt.axis('off')       # Hide axes
plt.show()
```

#### 4. Display in Grayscale

```python
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()
```

Once an image is loaded into a NumPy array, we can display it using `plt.imshow()`.  

- `plt.imshow(img)` → Displays the image.  
- `cmap='gray'` → Converts the image to grayscale.  
- `plt.axis('off')` → Removes axis ticks for a cleaner view.  
- `plt.show()` → Renders the final output.  


#### 5. Using imshow() with Customization

```python
plt.imshow(img, cmap='viridis')   # Apply a colormap
plt.colorbar()                    # Show color scale
plt.show()
```

Matplotlib allows applying **colormaps** to images to visualize intensity variations.  

- `cmap='viridis'` → Applies the *Viridis* colormap (default in Matplotlib, smooth gradient from purple → yellow).  
- `plt.colorbar()` → Adds a color scale legend to the side, showing how pixel values map to colors.  
- `plt.show()` → Displays the final output.  


#### 6. Plot Multiple Images

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis('off')

axes[1].imshow(img, cmap='gray')
axes[1].set_title("Grayscale")
axes[1].axis('off')

plt.show()
```

The function `plt.subplots()` allows creating multiple plots in a single figure.  

- `fig, axes = plt.subplots(rows, cols, figsize=(width, height))`  
  - Creates a grid of subplots.  
  - `axes` is an array of subplot objects.  
- `imshow()` → Displays each image in its respective subplot.  
- `set_title()` → Adds a title to each subplot.  
- `axis('off')` → Hides the axes for a cleaner display.  

In this example:  
- Left subplot → Original image  
- Right subplot → Grayscale version of the same image  


## PART-B using tutorial sheet

```python
from PIL import Image

import matplotlib.pyplot as plt
import numpy as np
```

In this section, we combine **PIL (Pillow)** and **Matplotlib** for image processing and visualization.  

- `from PIL import Image` → Pillow library for opening, manipulating, and saving images.  
- `import matplotlib.pyplot as plt` → For plotting and visualization.  
- `import numpy as np` → For numerical operations and converting images into arrays.  


```python
img = np.asarray(Image.open('the-deer-bug-the-stag-beetle-insects-oak.jpg'))
print(repr(img))
```

Using **Pillow (PIL)** with **NumPy**, we can convert an image into an array for analysis.  

- `Image.open("filename")` → Opens the image file.  
- `np.asarray(image)` → Converts the image into a NumPy array.  
  - Each pixel is represented by RGB (or RGBA) values.  
- `print(repr(img))` → Prints a detailed representation of the array instead of rendering the image.  


```python
imgplot = plt.imshow(img)
```

Once the image is converted into a NumPy array, we can display it using `plt.imshow()`.  

- `plt.imshow(img)` → Displays the image stored in the array `img`.  
- Matplotlib automatically interprets the shape of the array:  
  - `(H, W)` → Grayscale  
  - `(H, W, 3)` → RGB  
  - `(H, W, 4)` → RGBA (with transparency)  
- The returned object `imgplot` is an `AxesImage` instance, which can be further customized.  


```python
lum_img = img[:, :, 0]
plt.imshow(lum_img)
```

Images in RGB format have 3 channels: Red, Green, and Blue.  

- `img[:, :, 0]` → Extracts the **Red channel** (first channel) of the image.  
- `img[:, :, 1]` → Green channel  
- `img[:, :, 2]` → Blue channel  

We can visualize a single channel to analyze intensity or for image processing purposes.


```python
plt.imshow(lum_img, cmap="hot")
```

Once a single channel (like Red) is extracted, we can apply a **colormap** to visualize intensity variations more clearly.  

- `cmap="hot"` → Displays low values as dark colors and high values as bright colors (useful for heatmaps).  
- `plt.imshow(lum_img, cmap="hot")` → Applies the colormap to the single-channel image.  


```python
imgplot = plt.imshow(lum_img)
imgplot.set_cmap('nipy_spectral')
```

Matplotlib allows changing the colormap of an already created image object.  

- `imgplot = plt.imshow(lum_img)` → Displays the image and returns an `AxesImage` object.  
- `imgplot.set_cmap('nipy_spectral')` → Changes the colormap of the displayed image to `'nipy_spectral'`.  
- `plt.show()` → Renders the image with the new colormap.  

This is useful if you want to dynamically change how the image is visualized without re-plotting it.


```python
imgplot = plt.imshow(lum_img)
plt.colorbar()
```

A **colorbar** shows the mapping between the pixel values of an image and the colors used to represent them.  

- `imgplot = plt.imshow(lum_img)` → Displays the image and returns an `AxesImage` object.  
- `plt.colorbar()` → Adds a vertical color scale alongside the image.  
- This is especially useful when visualizing intensity variations in single channels or heatmaps.


```python
plt.hist(lum_img.ravel(), bins=range(256), fc='k', ec='k')
```

Histograms are useful to visualize the **distribution of pixel values** in an image.  

- `lum_img.ravel()` → Flattens the 2D array into 1D so all pixel values can be analyzed.  
- `bins=range(256)` → Divides values into 256 bins (0–255 for 8-bit images).  
- `fc='k'` → Sets the **face color** of the bars to black.  
- `ec='k'` → Sets the **edge color** of the bars to black.  


The histogram shows that most of the pixels in the image have a mid-range brightness, indicated by the prominent peaks between pixel values 100 and 150. This means the image is well-exposed but lacks very dark shadows (values near 0) and very bright highlights (values near 255)

```python
plt.imshow(lum_img, clim=(0, 175))
```

The `clim` parameter controls the **color limits** when displaying an image with `imshow()`.  

- `clim=(vmin, vmax)` → Maps the minimum (`vmin`) and maximum (`vmax`) pixel values to the colormap range.  
- Pixels below `vmin` → appear as the lowest color  
- Pixels above `vmax` → appear as the highest color  

This is useful to enhance contrast or focus on a specific intensity range.


```python
imgplot = plt.imshow(lum_img)
imgplot.set_clim(0, 175)
```

Matplotlib allows adjusting the intensity range **after the image has been displayed** using the `AxesImage` object.  

- `imgplot = plt.imshow(lum_img)` → Displays the image and returns an `AxesImage` object.  
- `imgplot.set_clim(vmin, vmax)` → Sets the minimum (`vmin`) and maximum (`vmax`) pixel values for the colormap.  
- `plt.show()` → Renders the updated image with the new intensity limits.  

This method is useful when you want to dynamically adjust contrast without re-plotting the image.


```python
img = Image.open('the-deer-bug-the-stag-beetle-insects-oak.jpg')
img.thumbnail((64, 64))  # resizes image in-place
imgplot = plt.imshow(img)
```

Pillow (PIL) provides the `thumbnail()` method to **resize an image in-place** while maintaining its aspect ratio.  

- `Image.open("filename")` → Opens the image.  
- `img.thumbnail((width, height))` → Resizes the image so that it **fits within the given size** while keeping the aspect ratio.  
- `plt.imshow(img)` → Displays the resized image with Matplotlib.  


```python
imgplot = plt.imshow(img, interpolation="bilinear")
```

Interpolation smooths the display of an image when resizing or zooming.  

- `interpolation="bilinear"` → Uses bilinear interpolation: each pixel is a weighted average of its neighbors, producing a smoother image.  
- Other options:  
  - `"nearest"` → No interpolation, blocky/pixelated effect  
  - `"bicubic"` → Higher-order interpolation, smoother than bilinear  
  - `"lanczos"` → High-quality resampling  

Interpolation only affects **display**, not the underlying image data.


```python
imgplot = plt.imshow(img, interpolation="bicubic")
```

Bicubic interpolation provides **smoother results** than bilinear interpolation by using a higher-order polynomial to compute pixel values.  

- `interpolation="bicubic"` → Smooths the image when scaling or resizing.  
- Especially useful for small images being displayed larger, reducing blocky/pixelated appearance.  
- Other common interpolation methods: `"nearest"`, `"bilinear"`, `"lanczos"`.
