#  Arrays with NumPy

##  1. Getting Started with NumPy
`NumPy` stands for **Numerical Python** and is a fundamental package for scientific computing in Python.

---

###  Key Features
- **ndarray (N-Dimensional Array):**
  - Similar to Python lists but faster and more memory-efficient.
- **Vectorized Operations:**
  - Perform arithmetic on arrays without writing loops.
- **Math Utilities:**
  - Linear algebra, statistics, random number generation, etc.

---

###  Why prefer NumPy arrays over Python lists?
1. **Speed:**  
   NumPy arrays are stored in **contiguous memory blocks**, enabling faster processing.
2. **Convenience:**  
   You can apply operations directly on the entire array.
3. **Efficiency:**  
   Less memory usage compared to Python lists.

print("NumPy Array:", np_array + 10)


```python
# Importing NumPy
import numpy as np

# Comparing Python list vs NumPy array
python_list = [1, 2, 3, 4, 5]
numpy_array = np.array([1, 2, 3, 4, 5])

print("Python List:", python_list)
print("NumPy Array:", numpy_array)
```

## 2. Creating NumPy Arrays

There are multiple ways to create arrays in NumPy:

1. **From a Python list or tuple** using `np.array()`.
2. **Arrays filled with zeros or ones** using `np.zeros()` and `np.ones()`.
3. **Range of numbers** using `np.arange(start, stop, step)`.
4. **Evenly spaced numbers** using `np.linspace(start, stop, num_points)`.

These methods are essential for scientific computing and simulations.


```python
# 1. Creating array from list
arr1 = np.array([10, 20, 30, 40, 50])

# 2. Array of zeros
arr2 = np.zeros((3, 4))  # 3 rows, 4 columns

# 3. Array of ones
arr3 = np.ones((2, 5))  # 2 rows, 5 columns

# 4. Range of numbers
arr4 = np.arange(0, 20, 2)  # Numbers from 0 to 18 with step size 2

# 5. Evenly spaced numbers
arr5 = np.linspace(0, 1, 5)  # 5 numbers between 0 and 1

print("Array from List:", arr1)
print("Zeros Array:\n", arr2)
print("Ones Array:\n", arr3)
print("Range Array:", arr4)
print("Linspace Array:", arr5)
```

## 3. Array Attributes

Each NumPy array has important attributes that describe its structure:

- **`ndim`** → Number of dimensions (axes).  
- **`shape`** → Tuple representing size along each dimension.  
- **`size`** → Total number of elements in the array.  
- **`dtype`** → Data type of array elements.  

These attributes are useful to quickly understand the structure of the data.


```python
# Creating a 2D array
arr = np.array([[1, 2, 3], [4, 5, 6]])

# Checking array attributes
print("Array:\n", arr)
print("Number of Dimensions (ndim):", arr.ndim)
print("Shape (rows, cols):", arr.shape)
print("Total Elements (size):", arr.size)
print("Data Type (dtype):", arr.dtype)
```

## 4. Array Indexing and Slicing

Just like Python lists, we can access individual elements and slices of arrays:

- **1D Arrays:** Use `[index]` and `[start:end:step]`.  
- **2D Arrays:** Use `[row, col]` for single elements.  
- Slicing allows extracting sub-arrays, rows, or columns.


```python
# 1D Array
arr1d = np.array([10, 20, 30, 40, 50])
print("1D Array:", arr1d)
print("Element at index 2:", arr1d[2])     # third element
print("Slice [1:4]:", arr1d[1:4])          # elements from index 1 to 3

# 2D Array
arr2d = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("\n2D Array:\n", arr2d)

# Accessing single element
print("Element at row 1, col 2:", arr2d[1, 2])

# Extracting a row
print("Second row:", arr2d[1, :])

# Extracting a column
print("First column:", arr2d[:, 0])

# Sub-matrix
print("Sub-matrix (first 2 rows, first 2 cols):\n", arr2d[0:2, 0:2])
```

## 5. Array Operations

NumPy arrays support a wide variety of operations:

1. **Elementwise Arithmetic**
   - Addition, subtraction, multiplication, division.
   - Operations are applied to each element.

2. **Broadcasting**
   - NumPy allows operations between arrays of different shapes if they are compatible.
   - Example: Adding a scalar to an array automatically applies the operation to all elements.

3. **Universal Functions (ufuncs)**
   - Optimized functions that work elementwise on arrays.
   - Examples: `np.sqrt`, `np.exp`, `np.log`, `np.sin`, `np.cos`.

4. **Comparison Operators**
   - Returns boolean arrays (`True`/`False`) for conditions.
   - Useful for **boolean indexing** and filtering.


```python
# Creating two arrays
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

# 1. Elementwise arithmetic
print("Addition:", a + b)
print("Subtraction:", b - a)
print("Multiplication:", a * b)
print("Division:", b / a)

# 2. Broadcasting (adding scalar)
print("\nBroadcasting (add 5):", a + 5)

# 3. Universal functions
print("\nSquare Root:", np.sqrt(a))
print("Exponential:", np.exp(a))
print("Sine Values:", np.sin(a))

# 4. Comparison operators
print("\nComparison (a > 2):", a > 2)
print("Comparison (b == 20):", b == 20)

# Boolean indexing
print("Elements of a greater than 2:", a[a > 2])
```

## 6. Reshaping and Flattening

NumPy allows changing the shape of arrays without altering their data.

1. **Reshaping**
   - Use `.reshape(rows, cols)` to change the structure.
   - The number of elements must remain the same.

2. **Flattening**
   - Convert multi-dimensional arrays into 1D.
   - Methods:
     - `.ravel()` → Returns a **view** (changes reflect in original array).
     - `.flatten()` → Returns a **copy** (changes do not affect original).


```python
# Creating a sequential array
arr = np.arange(1, 13)  # numbers from 1 to 12
print("Original Array:", arr)

# 1. Reshape into 3x4 matrix
reshaped = arr.reshape(3, 4)
print("\nReshaped to 3x4:\n", reshaped)

# 2. Flattening using ravel
ravelled = reshaped.ravel()
print("\nFlattened using ravel:", ravelled)

# 3. Flattening using flatten
flattened = reshaped.flatten()
print("Flattened using flatten:", flattened)

# Modifying ravel result affects original
ravelled[0] = 99
print("\nAfter modifying ravelled array:")
print("Ravelled:", ravelled)
print("Original reshaped array:\n", reshaped)

# Modifying flatten result does NOT affect original
flattened[1] = 88
print("\nAfter modifying flattened array:")
print("Flattened:", flattened)
print("Original reshaped array remains unchanged:\n", reshaped)
```

## 7. Statistical and Mathematical Functions

NumPy provides many built-in functions to perform **statistics and mathematics** efficiently:

### Common Statistical Functions
- **`np.mean()`** → Average value.  
- **`np.median()`** → Middle value.  
- **`np.std()`** → Standard deviation (spread of data).  
- **`np.var()`** → Variance.  
- **`np.sum()`** → Sum of all elements.  
- **`np.min()` / `np.max()`** → Minimum and maximum values.  

### Axis Parameter
- **`axis=0`** → Apply function column-wise (down each column).  
- **`axis=1`** → Apply function row-wise (across each row).  


```python
# Creating a 2D array
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:\n", arr)

# 1. Basic statistics
print("\nMean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))
print("Sum:", np.sum(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

# 2. Using axis
print("\nColumn-wise Mean (axis=0):", np.mean(arr, axis=0))
print("Row-wise Sum (axis=1):", np.sum(arr, axis=1))
```

## 8. Random Module in NumPy

NumPy has a powerful random module (`np.random`) to generate random numbers for simulations, testing, and machine learning.

### Common Functions:
1. **`np.random.rand(d0, d1, ...)`**
   - Generates random numbers in the range `[0, 1)`.
   - Uniform distribution.

2. **`np.random.randn(d0, d1, ...)`**
   - Generates numbers from the **standard normal distribution** (mean = 0, std = 1).

3. **`np.random.randint(low, high, size)`**
   - Generates random integers between `low` and `high`.

4. **`np.random.seed(value)`**
   - Fixes the randomness → useful for reproducibility.
   - Same seed produces the same random numbers.


```python
# 1. Random numbers between 0 and 1
rand_array = np.random.rand(3, 3)
print("Random array (0 to 1):\n", rand_array)

# 2. Random numbers from normal distribution
randn_array = np.random.randn(2, 4)
print("\nRandom array (normal distribution):\n", randn_array)

# 3. Random integers
randint_array = np.random.randint(1, 100, size=(5,))
print("\nRandom integers between 1 and 100:", randint_array)

# 4. Using seed for reproducibility
np.random.seed(42)
print("\nWith Seed 42, random numbers:", np.random.rand(3))
np.random.seed(42)
print("With Seed 42 again, same numbers:", np.random.rand(3))
```

## 9. Practical Example – Student Marks Analysis

Let’s combine multiple NumPy concepts in a small project.  
We will:

1. **Generate random student marks** for 5 subjects and 10 students.
2. **Analyze the marks** using statistical functions.
3. **Reshape and slice** data for specific insights.
4. **Apply boolean indexing** to find top-performing students.

This demonstrates how NumPy is useful in real-world data analysis.

```python
# Step 1: Generate random marks (10 students, 5 subjects)
np.random.seed(101)  # reproducibility
marks = np.random.randint(40, 100, size=(10, 5))  # marks between 40 and 100

print("Marks of 10 students (rows) across 5 subjects (cols):\n", marks)

# Step 2: Analyze data
print("\nOverall Average Marks:", np.mean(marks))
print("Average Marks per Subject:", np.mean(marks, axis=0))
print("Average Marks per Student:", np.mean(marks, axis=1))

# Step 3: Find highest and lowest scoring student
total_marks = np.sum(marks, axis=1)
highest_student = np.argmax(total_marks)  # index of top student
lowest_student = np.argmin(total_marks)   # index of bottom student

print("\nTotal Marks of Each Student:", total_marks)
print("Top Student Index:", highest_student, "with Total Marks:", total_marks[highest_student])
print("Lowest Student Index:", lowest_student, "with Total Marks:", total_marks[lowest_student])

# Step 4: Boolean indexing - Students with average > 70
avg_per_student = np.mean(marks, axis=1)
print("\nStudents with Avg > 70:", np.where(avg_per_student > 70)[0])
```
